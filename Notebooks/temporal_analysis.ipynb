# Temporal Analysis of Robinson Crusoe Adaptations (1719-1903)

Analyze how adaptations evolved over time:
- Distribution of adaptations across decades
- Stylistic drift over centuries
- Embedding similarity trends over time
- Model performance by publication era

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import tensorflow_hub as hub
from sklearn.metrics.pairwise import cosine_similarity
import warnings
warnings.filterwarnings('ignore')

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (14, 8)

print("Environment ready")
```

## 1. Load Metadata

```python
# Load Robinson Crusoe metadata
rc_metadata = pd.read_csv('../Texts/ROBINSON_CRUSOE_TEXTS/RC.csv')

print(f"Metadata: {len(rc_metadata)} entries")
print(f"\nColumns: {rc_metadata.columns.tolist()}")
print(f"\nSample:")
print(rc_metadata.head())

# Clean year data
rc_metadata['year'] = pd.to_numeric(rc_metadata['year'], errors='coerce')
rc_metadata = rc_metadata.dropna(subset=['year'])
rc_metadata['year'] = rc_metadata['year'].astype(int)

print(f"\nYear range: {rc_metadata['year'].min()} - {rc_metadata['year'].max()}")
```

## 2. Temporal Distribution

```python
# Create decade bins
rc_metadata['decade'] = (rc_metadata['year'] // 10) * 10

# Count by decade
decade_counts = rc_metadata['decade'].value_counts().sort_index()

print("\nAdaptations by Decade:")
print("=" * 50)
for decade, count in decade_counts.items():
    print(f"{decade}s: {count:3d} adaptations")

# Visualize distribution
fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# Bar chart by decade
axes[0].bar(decade_counts.index, decade_counts.values, width=8, alpha=0.7, color='steelblue')
axes[0].set_xlabel('Decade', fontsize=12)
axes[0].set_ylabel('Number of Adaptations', fontsize=12)
axes[0].set_title('Robinson Crusoe Adaptations Over Time', fontsize=14, fontweight='bold')
axes[0].grid(axis='y', alpha=0.3)

# Cumulative count
cumulative = decade_counts.sort_index().cumsum()
axes[1].plot(cumulative.index, cumulative.values, marker='o', linewidth=2, markersize=6)
axes[1].fill_between(cumulative.index, cumulative.values, alpha=0.3)
axes[1].set_xlabel('Decade', fontsize=12)
axes[1].set_ylabel('Cumulative Adaptations', fontsize=12)
axes[1].set_title('Cumulative Growth of Adaptations', fontsize=14, fontweight='bold')
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('temporal_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

print("\nVisualization saved as 'temporal_distribution.png'")
```

## 3. Language Distribution Over Time

```python
# Analyze language distribution
if 'language' in rc_metadata.columns:
    lang_by_period = rc_metadata.groupby(['decade', 'language']).size().unstack(fill_value=0)
    
    # Plot stacked bar chart
    fig, ax = plt.subplots(figsize=(14, 8))
    lang_by_period.plot(kind='bar', stacked=True, ax=ax, alpha=0.8)
    ax.set_xlabel('Decade', fontsize=12)
    ax.set_ylabel('Number of Adaptations', fontsize=12)
    ax.set_title('Language Distribution of Adaptations Over Time', fontsize=14, fontweight='bold')
    ax.legend(title='Language', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig('language_temporal.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print("Visualization saved as 'language_temporal.png'")
else:
    print("Language information not available in metadata")
```
